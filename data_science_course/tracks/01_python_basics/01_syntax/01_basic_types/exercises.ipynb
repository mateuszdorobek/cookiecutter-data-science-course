# Variables and Operators Exercises


## 1. Variables Initialization


### Exercise 1
Initialize variables to store your name, age, and favorite color. Print them in a sentence.
Expected Output: "My name is Alex, I am 25 years old, and my favorite color is blue."


### Exercise 2
Assign the radius of a circle to a variable. Calculate and print its area using `pi = 3.14`.
Expected Output: For radius 5, the area is 78.5.


### Exercise 3
Swap the values of two variables without using a third variable.
Example: x = 5, y = 10. After swapping, x = 10, y = 5.


### Exercise 4
Using multi-variable assignment, assign the values 1, 2, and 3 to variables `a`, `b`, and `c`. Print their sum.


## 2. Arithmetic Operators


### Exercise 5
Write a program to calculate the area of a circle based on a radius entered by the user.
Use the formula: `area = pi * r^2`. Ensure `pi = 3.14159`.
Expected Output: "The area of the circle with radius 5 is 78.54."


### Exercise 6
Given two points (x1, y1) and (x2, y2), calculate the distance between them.
Formula: `distance = sqrt((x2 - x1)^2 + (y2 - y1)^2)`.
Expected Output: "The distance between (1, 1) and (4, 5) is 5.0."


### Exercise 7
Compute the area of intersection for two rectangles given their corner coordinates.
Use the formula: `x_overlap * y_overlap`.
Expected Output: For overlapping rectangles, print the area of the intersection.


## 3. Comparison Operators


### Exercise 8
Write a program to compare two numbers and print whether the first is greater, less, or equal to the second.
Example: Input: 10, 20. Output: "10 is less than 20."


### Exercise 9
Check if a given number is within a specific range (e.g., between 1 and 100).


## 4. Logical Operators


### Exercise 10
Write a program that checks if a given year is a leap year using logical operators.
Expected Output: "2000 is a leap year." or "1900 is not a leap year."


### Exercise 11
Write a program to validate a password based on the following rules:
- It should be at least 8 characters long.
- It should contain at least one uppercase letter, one lowercase letter, and one digit.


## 5. Assignment Operators


### Exercise 12
Write a program to calculate compound interest using the formula:
`A = P * (1 + r/n)^(n*t)`.
Assign initial values to `P`, `r`, `n`, and `t`. Calculate and print `A`.


### Exercise 13
Write a program to increment a variable `x` from 1 to 10 using `+=` and print each value.


## 6. Advanced Challenges


### Exercise 14
Write a program to convert temperatures between Celsius and Fahrenheit.
Formulas:
- `F = (C * 9/5) + 32`
- `C = (F - 32) * 5/9`
Expected Output:
- "60°C is 140.0°F"
- "45°F is 7.0°C"


### Exercise 15
Write a program that validates whether a given point lies inside, outside, or on the boundary of a circle.
The circle is centered at (0, 0) with radius `r`.


### Exercise 16
Calculate the BMI (Body Mass Index) and classify it using the following ranges:
- Underweight: < 18.5
- Normal weight: 18.5–24.9
- Overweight: 25–29.9
- Obesity: >= 30
Formula: `BMI = weight / height^2`
